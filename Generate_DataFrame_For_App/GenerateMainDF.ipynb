```python
# 2027 -- prognosticated

df_2027 = pd.read_csv(stars_dir + "2027/predictions_From_May_2026/R_CSV_output/Star_2026may.csv")

df_2027['PROVIDER_ID'] = df_2027['PROVIDER_ID'].astype(str)
df_2027 = df_2027.applymap(lambda x: x.decode() if isinstance(x, bytes) else x)
df_2027.columns = df_2027.columns.str.strip()

## Replace the imputed 666666 suffixes of VHA hospitals with their original 'F' suffix
prvdrs = []
for p in df_2027['PROVIDER_ID'].tolist():
    p = str(p)
    if '666666' in p:
        p = p[:-6]
        p = p + 'F'
    while len(p) < 6:
        p = '0' + p
    prvdrs.append(p)
    
df_2027['PROVIDER_ID'] = prvdrs

for i in list(df_2027):
    if i == 'cnt_grp':
        df_2027[i] = df_2027[i].astype(str).str[0]
    if i == 'PROVIDER_ID':
        pass
    else:
        df_2027[i] = pd.to_numeric(df_2027[i], errors='coerce')

df_2027['Release year'] = ['2027']*df_2027.shape[0]
df_2027 = df_2027[~df_2027['star'].isin([np.nan, float("NaN")])]
prvdrs_2027 = df_2027['PROVIDER_ID'].unique()

print(len(prvdrs_2027), 'hospitals with stars predictions for 2027')
for star in [1,2,3,4,5]:
    tdf = df_2027[df_2027['star'] == star]
    print(tdf.shape[0], str(star)+'-star hospitals')

df_2027.head()
```

## Load SAS pack secondary (domain-specific) files
These data files pertain to standardized scores for each component of a particular measure domain.

```python
# 2027 -- predicted

fnames = ['outcome_readmission', 'outcome_safety', 'process', 'ptexp']
f_ext = ['readmission_', 'safety_', 'process_', 'patient_exp_']

with io.capture_output() as captured: 
    df_2027 = pd.read_csv(stars_dir + '2027/predictions_From_May_2026/R_CSV_output/outcome_mortality.csv')
    #df_2027 = df_2027.applymap(lambda x: x.decode() if isinstance(x, bytes) else x)
    
    ls = []
    for p in df_2027['PROVIDER_ID'].tolist():
        p = int(p)
        p = str(p)
        if '666666' in p:
            p = p[:-6]
            p = p + 'F'
        while len(p) < 6:
            p = '0' + p
        ls.append(p)
    df_2027['PROVIDER_ID'] = ls
    

    
cols = list(df_2027)
new_cols = []
ind = cols.index('total_cnt')
for i, c in enumerate(cols):
    if i >= ind:
        c = 'mortality_' + c
    new_cols.append(c)
print(new_cols)
df_2027.columns = new_cols

for j, f in enumerate(fnames):
    with io.capture_output() as captured: 
        tdf = pd.read_csv(stars_dir + '2027/predictions_From_May_2026/R_CSV_output/' + f + '.csv')
    #tdf = tdf.applymap(lambda x: x.decode() if isinstance(x, bytes) else x)
    
    ls = []
    for p in tdf['PROVIDER_ID'].tolist():
        p = int(p)
        p = str(p)
        if '666666' in p:
            p = p[:-6]
            p = p + 'F'
        while len(p) < 6:
            p = '0' + p
        ls.append(p)
    tdf['PROVIDER_ID'] = ls
    
    cols = list(tdf)
    new_cols = []
    ind = cols.index('total_cnt')
    for i, c in enumerate(cols):
        if i >= ind:
            c = f_ext[j] + c
        new_cols.append(c)
    tdf.columns = new_cols

    df_2027 = df_2027.merge(tdf, on='PROVIDER_ID', how='outer')
    
df_2027['Release year'] = ['2027']*df_2027.shape[0]
df_2027 = df_2027.applymap(lambda x: x.decode() if isinstance(x, bytes) else x)
df_2027 = df_2027[df_2027['PROVIDER_ID'].isin(prvdrs_2027)]
prvdrs = df_2027['PROVIDER_ID'].unique()

print(len(prvdrs), 'hospitals in 2027')
df_2027.head()
```
